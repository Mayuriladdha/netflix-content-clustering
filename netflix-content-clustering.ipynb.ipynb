# **Project Summary -**

I analyzed the Netflix Movies and TV Shows dataset to ensure it aligns with the project's objective of clustering similar content using unsupervised machine learning techniques. The dataset contains information about movies and TV shows available on Netflix, including attributes such as title, type, director, cast, country, release year, rating, duration, genres, and descriptions. Like many real-world datasets, it contains missing values, duplicate records, and categorical features that require preprocessing before applying machine learning algorithms.

The project began with an in-depth exploratory data analysis (EDA) to understand the structure of the dataset and identify important trends. Various visualizations were used to analyze the distribution of content types, release years, ratings, countries, genres, and durations. The analysis revealed that movies constitute a larger portion of Netflix's catalog than TV shows, while content production has grown significantly over recent years. Popular genres such as International Movies, Dramas, and Comedies dominate the platform, reflecting the diverse preferences of Netflix's global audience.

After understanding the data, the preprocessing stage involved handling missing values, removing unnecessary columns, eliminating duplicates, and transforming categorical and textual information into a suitable format for machine learning. Since the primary objective was to cluster similar content based on textual characteristics, the text data was cleaned by converting it to lowercase, removing punctuation, stop words, and extra spaces. Relevant textual features such as genres, descriptions, directors, cast members, ratings, and countries were combined to create a comprehensive representation of each title.

To convert the processed text into numerical form, the TF-IDF (Term Frequency–Inverse Document Frequency) vectorization technique was applied. This approach assigns weights to important words while reducing the influence of commonly occurring terms, enabling a meaningful representation of the content. Because the resulting feature space is high-dimensional, dimensionality reduction techniques were employed to improve computational efficiency and enhance clustering performance.

The K-Means clustering algorithm was then applied to group similar Netflix titles into distinct clusters. The optimal number of clusters was determined using evaluation techniques such as the Elbow Method and Silhouette Score, ensuring a balance between cluster compactness and separation. Once the clusters were formed, they were analyzed to identify common characteristics among the grouped titles. Each cluster represented content with similar themes, genres, descriptions, or audience categories, making it easier to understand patterns within Netflix's extensive content library.

The results demonstrate that unsupervised machine learning can effectively organize large collections of entertainment content without requiring predefined labels. Such clustering techniques can support recommendation systems, improve content categorization, and help streaming platforms better understand their content inventory. Overall, this project showcases a complete machine learning workflow, including data preprocessing, exploratory analysis, feature engineering, text vectorization, clustering, evaluation, and interpretation. The insights gained from the analysis highlight how data-driven approaches can contribute to enhancing user experience and strategic decision-making in digital streaming services.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Create a copy of the dataset
df1 = df.copy()

# Remove duplicate records
df1.drop_duplicates(inplace=True)

# Check missing values
df1.isnull().sum()

# Fill missing values
df1['director'].fillna('Unknown', inplace=True)
df1['cast'].fillna('Unknown', inplace=True)
df1['country'].fillna('Unknown', inplace=True)
df1['rating'].fillna(df1['rating'].mode()[0], inplace=True)
df1['date_added'].fillna(df1['date_added'].mode()[0], inplace=True)

# Convert 'date_added' to datetime format
df1['date_added'] = pd.to_datetime(df1['date_added'])

# Verify the dataset
df1.info()
df1.head()
````

### What all manipulations have you done and insights you found?

To prepare the dataset for analysis and clustering, several preprocessing steps were performed. First, a copy of the original dataset was created to preserve the raw data. Duplicate records were identified and removed to eliminate redundancy and ensure that each observation represented a unique Netflix title. Missing values were then examined, and appropriate strategies were used to handle them. The director, cast, and country columns were filled with "Unknown" since these are categorical features, while missing values in rating and date_added were replaced using their most frequent values (mode). Additionally, the date_added column was converted to the datetime format to facilitate time-based analysis.

These preprocessing steps improved the overall quality and consistency of the dataset, making it suitable for exploratory data analysis and machine learning. During this process, it was observed that missing values were concentrated mainly in the director, cast, country, date_added, and rating columns, while the remaining features were complete. Duplicate records were relatively few but were removed to prevent biased results. After preprocessing, the dataset became clean, consistent, and ready for feature engineering, visualization, and clustering analysis.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Hypothetical Statement - 1

Research Hypothesis:
There is a significant relationship between the type of content (Movies and TV Shows) and content ratings on Netflix.

Null Hypothesis (H₀): There is no significant relationship between the type of content (Movies and TV Shows) and content ratings on Netflix. The two variables are independent.

Alternative Hypothesis (H₁): There is a significant relationship between the type of content (Movies and TV Shows) and content ratings on Netflix. The two variables are dependent.

#### 2. Perform an appropriate statistical test.

````python
import pandas as pd
from scipy.stats import chi2_contingency

# Create contingency table
contingency_table = pd.crosstab(df1['type'], df1['rating'])

# Perform Chi-Square Test
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

# Display results
print("Chi-Square Statistic:", chi2)
print("Degrees of Freedom:", dof)
print("P-Value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis (H₀)")
    print("There is a significant relationship between content type and rating.")
else:
  print("Fail to Reject the Null Hypothesis (H₀)")
  print("There is no significant relationship between content type and rating.")
````

##### Which statistical test have you done to obtain P-Value?

I performed the Chi-Square Test of Independence to obtain the P-value. This test is appropriate because both variables—content type (Movies and TV Shows) and content rating—are categorical. The Chi-Square Test determines whether there is a statistically significant association between these two variables in the Netflix dataset.

##### Why did you choose the specific statistical test?


I chose the Chi-Square Test of Independence because both the variables involved—content type (Movies and TV Shows) and content rating—are categorical variables. The Chi-Square Test is specifically designed to determine whether there is a statistically significant association between two categorical variables. It compares the observed frequencies with the expected frequencies under the assumption of independence, making it the most appropriate test for this hypothesis.

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Hypothetical Statement - 2

Research Hypothesis:

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
# Import required library
from scipy.stats import ttest_ind

# Separate release years for Movies and TV Shows
movies = df1[df1['type'] == 'Movie']['release_year'].dropna()
tv_shows = df1[df1['type'] == 'TV Show']['release_year'].dropna()

# Perform Independent Two-Sample t-test
t_stat, p_value = ttest_ind(movies, tv_shows, equal_var=False)

# Display results
print("T-Statistic:", t_stat)
print("P-Value:", p_value)



alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis (H₀)")
    print("There is a significant difference in the release years of Movies and TV Shows.")
else:
    print("Fail to Reject the Null Hypothesis (H₀)")
    print("There is no significant difference in the release years of Movies and TV Shows.")
````

Research Hypothesis:
There is a significant difference in the duration of Movies and TV Shows available on Netflix.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
# Import required libraries
import pandas as pd
from scipy.stats import mannwhitneyu

# Create a copy of the dataset
duration_df = df1.copy()

# Extract numeric duration values
duration_df['duration_numeric'] = duration_df['duration'].str.extract('(\d+)').astype(float)

# Separate Movie and TV Show durations
movie_duration = duration_df[duration_df['type'] == 'Movie']['duration_numeric'].dropna()
tv_duration = duration_df[duration_df['type'] == 'TV Show']['duration_numeric'].dropna()

# Perform Mann-Whitney U Test
u_stat, p_value = mannwhitneyu(movie_duration, tv_duration, alternative='two-sided')

# Display results
print("Mann-Whitney U Statistic:", u_stat)
print("P-Value:", p_value)





alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis (H₀)")
    print("There is a significant difference between the duration of Movies and TV Shows.")
else:
    print("Fail to Reject the Null Hypothesis (H₀)")
    print("There is no significant difference between the duration of Movies and TV Shows.")
````

##### Why did you choose the specific statistical test?

I chose the Mann–Whitney U Test because it is suitable for comparing two independent groups when the data may not be normally distributed. ### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# Extract numeric duration from the duration column
df1['duration_numeric'] = df1['duration'].str.extract('(\d+)').astype(float)


# Select numerical columns
numeric_cols = ['release_year', 'duration_numeric']
df1[numeric_cols].describe()

# Visualize outliers using boxplots
plt.figure(figsize=(12,5))

for i, col in enumerate(numeric_cols):
    plt.subplot(1, 2, i+1)
    sns.boxplot(y=df1[col])
    plt.title(f'Boxplot of {col}')

plt.tight_layout()
plt.show()

# Function to remove outliers using IQR method
def remove_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return data[(data[column] >= lower) & (data[column] <= upper)]

# Apply IQR method
df1 = remove_outliers_iqr(df1, 'release_year')
df1 = remove_outliers_iqr(df1, 'duration_numeric')




# Verify outlier removal
plt.figure(figsize=(12,5))

for i, col in enumerate(numeric_cols):
    plt.subplot(1, 2, i+1)
    sns.boxplot(y=df1[col])
    plt.title(f'Boxplot After Outlier Removal ({col})')

plt.tight_layout()
plt.show()
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

The Interquartile Range (IQR) method was used to detect and treat outliers in the numerical features, such as release_year and duration_numeric. The IQR method identifies outliers by calculating the first quartile (Q1), third quartile (Q3), and the interquartile range (IQR = Q3 − Q1). Values lying below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR were considered outliers and removed.

This technique was chosen because it is simple, robust, and does not assume that the data follows a normal distribution. It is less affected by extreme values than methods based on the mean and standard deviation, making it well-suited for the Netflix dataset. Removing these outliers helped reduce noise, improved data quality, and ensured that the clustering algorithm produced more reliable and meaningful groups.

````python
# Expand Contraction

# Install contractions package (Run only once if not installed)
!pip install contractions

# Import library
import contractions


# Function to expand contractions
def expand_contractions(text):
    if pd.isna(text):
            return ""
    return contractions.fix(text)

# Apply contraction expansion on the description column
df1['description'] = df1['description'].apply(expand_contractions)


# Display sample results
df1[['description']].head()
````

````python
# Lower Casing
# Convert text to lowercase
df1['description'] = df1['description'].str.lower()

# Display sample output
df1[['description']].head()
````

#### 3. Removing Punctuations

````python
# Remove Punctuations
# Import regular expression library
import re


# Function to remove punctuations
def remove_punctuation(text):
    if pd.isna(text):
            return ""
    return re.sub(r'[^\w\s]', '', text)


# Apply punctuation removal to the description column
df1['description'] = df1['description'].apply(remove_punctuation)



# Display sample output
df1[['description']].head()
````

#### 4. Removing URLs & Removing words and digits contain digits.

````python
# Remove URLs & Remove words and digits contain digits

# Function to remove URLs
def remove_urls(text):
    if pd.isna(text):
            return ""
    return re.sub(r'http\S+|www\S+|https\S+', '', text)



# Apply URL removal
df1['description'] = df1['description'].apply(remove_urls)


# Function to remove words containing digits
def remove_words_with_digits(text):
    if pd.isna(text):
            return ""
    return re.sub(r'\b\w*\d\w*\b', '', text)



# Apply removal of words containing digits
df1['description'] = df1['description'].apply(remove_words_with_digits)




# Display sample output
df1[['description']].head()
````

#### 5. Removing Stopwords & Removing White spaces

````python
# Remove Stopwords


# Import NLTK stopwords
import nltk
from nltk.corpus import stopwords

# Download stopwords (Run only once)
nltk.download('stopwords')



# Define stopwords
stop_words = set(stopwords.words('english'))
````

````python
# Remove White spaces


# Function to remove stopwords and extra white spaces
def clean_text(text):
    if pd.isna(text):
            return ""

   # Remove stopwords
    words = [word for word in text.split() if word not in stop_words]

   # Remove extra white space
    return " ".join(words).strip()


# Apply the function to the description column
df1['description'] = df1['description'].apply(clean_text)

# Display sample output
df1[['description']].head()
````

#### 6. Rephrase Text

````python
# Rephrase Text

# Function to normalize text
def normalize_text(text):
    if pd.isna(text):
            return ""
    return " ".join(text.split())

# Apply text normalization
df1['description'] = df1['description'].apply(normalize_text)

 # Display sample output
df1[['description']].head()
````

#### 7. Tokenization

````python
import nltk
from nltk.tokenize import word_tokenize

# Download tokenizer (Run only once)
nltk.download('punkt')
nltk.download('punkt_tab')


# Function for tokenization
def tokenize_text(text):
    if pd.isna(text):
            return []
    return word_tokenize(text)


# Apply tokenization to the description column
df1['tokens'] = df1['description'].apply(tokenize_text)



# Display sample output
df1[['description', 'tokens']].head()
````

#### 8. Text Normalization

````python
# Normalizing Text (i.e., Stemming, Lemmatization etc.)
# Import required libraries
import nltk
from nltk.stem import WordNetLemmatizer

# Download WordNet resources (Run only once)
nltk.download('wordnet')
nltk.download('omw-1.4')


# Initialize the lemmatizer
lemmatizer = WordNetLemmatizer()


# Function for lemmatization
def lemmatize_text(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]




# Apply lemmatization to the tokenized text
df1['lemmatized_tokens'] = df1['tokens'].apply(lemmatize_text)


# Join the lemmatized tokens into a single string
df1['clean_description'] = df1['lemmatized_tokens'].apply(lambda x: " ".join(x))



# Display sample output
df1[['description', 'clean_description']].head()
````

##### Which text normalization technique have you used and why?

Explanation of the ML Model

K-Means Clustering is an unsupervised machine learning algorithm that groups similar data points into a predefined number of clusters based on feature similarity. In this project, the model was trained using TF-IDF vectorized Netflix content descriptions to identify groups of similar movies and TV shows.

The model's performance was evaluated using the Silhouette Score, which measures the quality of clustering by assessing how similar each data point is to its own cluster compared to other clusters. A higher Silhouette Score indicates better-defined and well-separated clusters. The score was calculated for different values of K, and the optimal number of clusters was selected based on the highest Silhouette Score, resulting in more meaningful and interpretable content clusters.

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ML Model - 1 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)
# Import required libraries
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Hyperparameter tuning for K-Means
best_score = -1
best_k = 2
best_model = None

for k in range(2, 11):
    model = KMeans(n_clusters=k, random_state=42)
    labels = model.fit_predict(tfidf_matrix)
    score = silhouette_score(tfidf_matrix, labels)
    print(f"K = {k}, Silhouette Score = {score:.4f}")

    if score > best_score:
       best_score = score
       best_k = k
       best_model = model

print("\nBest Number of Clusters:", best_k)
print("Best Silhouette Score:", best_score)


# Fit the best model
best_model.fit(tfidf_matrix)




# Predict cluster labels
df1['Cluster'] = best_model.predict(tfidf_matrix)

# Display first few records
df1[['title', 'Cluster']].head()
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
import numpy as np


# Hyperparameter tuning using Silhouette Score

eps_values = [0.3, 0.5, 0.7, 1.0]
min_samples_values = [3, 5, 7]

best_score = -1
# Initialize best_eps and best_min_samples with the first values as a fallback.
# This ensures they are always valid floats/ints, preventing the InvalidParameterError
# if no combination results in more than one cluster.
best_eps = eps_values[0]
best_min_samples = min_samples_values[0]

for eps in eps_values:
    for min_samples in min_samples_values:

        dbscan = DBSCAN(
            eps=eps,
            min_samples=min_samples,
            metric='cosine'
        )

        labels = dbscan.fit_predict(tfidf_matrix)

        # Ignore noise points
        mask = labels != -1

        # Calculate score only if more than one cluster exists
        if len(set(labels[mask])) > 1:
            score = silhouette_score(tfidf_matrix[mask], labels[mask])
            print(f"eps={eps}, min_samples={min_samples}, Silhouette Score={score:.4f}")
            if score > best_score:
                best_score = score
                best_eps = eps
                best_min_samples = min_samples
        else:
            # Optionally, print a message if no valid clusters were formed for a given combination
            # This helps in debugging if all combinations fail.
            print(f"eps={eps}, min_samples={min_samples}: No meaningful clusters found (only noise or one cluster).")

print("Best eps:", best_eps)
print("Best min_samples:", best_min_samples)
print("Best Silhouette Score:", best_score)




# Train DBSCAN with the best hyperparameters
best_dbscan = DBSCAN(
    eps=best_eps,
    min_samples=best_min_samples,
    metric='cosine'
)
best_dbscan.fit(tfidf_matrix)





# Predict cluster labels
dbscan_labels = best_dbscan.fit_predict(tfidf_matrix)

# Add cluster labels to the dataframe
df1['DBSCAN_Cluster'] = dbscan_labels

# Display results
df1[['title', 'DBSCAN_Cluster']].head()
````

Different combinations of the eps (maximum neighborhood distance) and min_samples (minimum number of points required to form a cluster) parameters were evaluated, and the combination with the highest Silhouette Score was selected as the optimal model.

This technique was chosen because DBSCAN is an unsupervised learning algorithm, and therefore traditional hyperparameter optimization methods such as GridSearchCV, RandomizedSearchCV, and Bayesian Optimization are not suitable due to the absence of labeled target data. The Silhouette Score is an appropriate evaluation metric for clustering because it measures how well the data points are grouped within their own clusters while remaining distinct from other clusters.
